# Autoseed v2.2 Notebook
Uses autoseed and the C tools installed with the pip package.

```python
# PARAMETERS
INPUT_BG = "80bpshuffled.seq"
INPUT_SIGNAL = "STARRseq_promoter_80bp.seq"
min_seed = 5
max_seed = 8
top_seeds = 100
extend_k = 4
extend_top = 100
svg_prob = 0.20
OUTPUT_SVG = "autoseed_output.svg"
```

```python
# LOAD autoseed from the installed package (C tools are in the package bin/ folder)
from autoseed import core
from autoseed import run
print("Autoseed logic loaded.")
```

```python
# Example test cell: download example .seq files and run the standard Autoseed v2.2 pipeline
import urllib.request, pathlib, subprocess, sys

# Download example background and signal sequences from the autoseed GitHub repo if missing
base = "https://raw.githubusercontent.com/jutaipal/autoseed/main/"
for fname in [INPUT_BG, INPUT_SIGNAL]:
    p = pathlib.Path(fname)
    if not p.exists():
        print(f"Downloading {fname} ...")
        urllib.request.urlretrieve(base + fname, fname)
    else:
        print(f"Using existing {fname}")

# Run: autoseed 80bpshuffled.seq STARRseq_promoter_80bp.seq 5 8 100 4 100 0.35 1 0.20
args = [
    INPUT_BG,
    INPUT_SIGNAL,
    str(min_seed),
    str(max_seed),
    str(top_seeds),
    str(extend_k),
    str(extend_top),
    "0.35",
    "1",
    str(svg_prob),
]

print("Running autoseed with:", " ".join(args))
run(args)
```
